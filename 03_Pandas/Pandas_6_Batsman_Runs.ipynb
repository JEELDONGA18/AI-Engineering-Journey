### J. Mini Project Questions

```python
# Build a simple IPL Run Analysis Dashboard:

# Include:

# Top scorer
# Average runs
# Top 10 chart
# Distribution chart
# Performance categories

# This is built using ChatGPT
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Dataset
# df = pd.read_csv("batsman_runs_series.csv")

# ----------------------------
# Top Scorer
# ----------------------------

top_scorer = df.loc[df['batsman_run'].idxmax()]

print("Top Scorer:")
print(top_scorer)

# ----------------------------
# Average Runs
# ----------------------------

average_runs = df['batsman_run'].mean()

print("\nAverage Runs:", average_runs)

# ----------------------------
# Performance Category
# ----------------------------

def category(run):
    if run < 100:
        return "Low"
    elif run <= 500:
        return "Medium"
    else:
        return "High"

df['Category'] = df['batsman_run'].apply(category)

category_count = df['Category'].value_counts()

# ----------------------------
# Top 10 Players
# ----------------------------

top10 = df.sort_values(by='batsman_run',
                       ascending=False).head(10)

# ----------------------------
# Dashboard
# ----------------------------

fig, axes = plt.subplots(2, 2, figsize=(16,10))

# Top 10 Bar Chart
axes[0,0].bar(top10['batter'],
              top10['batsman_run'])

axes[0,0].set_title("Top 10 Run Scorers")
axes[0,0].tick_params(axis='x', rotation=45)

# Histogram
axes[0,1].hist(df['batsman_run'], bins=20)

axes[0,1].set_title("Runs Distribution")

# Pie Chart
axes[1,0].pie(category_count,
               labels=category_count.index,
               autopct='%1.1f%%')

axes[1,0].set_title("Performance Categories")

# Text Summary
axes[1,1].axis('off')

summary = f"""
Top Scorer:
{top_scorer['batter']}

Runs:
{top_scorer['batsman_run']}

Average Runs:
{average_runs:.2f}
"""

axes[1,1].text(0.1, 0.5, summary, fontsize=12)

# Dashboard Title
fig.suptitle("IPL Run Analysis Dashboard",
             fontsize=20)

plt.tight_layout()

plt.show()
```
