## 1. Indexing

```python
import pandas as pd
```

```python
s = pd.Series([10,20,30,40,50])
```

```python
s
```

```python
s[0] # access single element from series called INDEX ACCESS        --------> WAY 1
```

```python
s[0:2] # access perticular part from the series and this action is called as SLICING..           --------> WAY 2
        # start (included) : stop (excluded) : step (values to jump)
```

```python
#                                                                    --------> WAY 3 : iloc (Location based indexing)
s.iloc[2]
```

```python
s.iloc[[1,3,4]]  # pass all the locations as array of index to access that elements
```

```python
# Suppose given series represents the calories provided by the fruits 
s.name = "Calories"
index = ["Apple","Banana","Cherry","Mango","Watermelon"]
s.index = index
```

```python
s     # So here we provided the perticular indexes as fruit names instead of the default indices as 0,1,2,...
```

```python
s['Cherry']  # indexing element by new index name and this gives 30 calories are provided by the Cherry..
```

### Question : Can we use iloc here..??

```python
s.iloc['Cherry']
```

```python
# We got this error because iloc is only used for the Numeric indices like..
s.iloc[2]
```

### Then what is the solution ?? --> Answer : loc function 

```python
#                                                                    --------> WAY 4 : loc (Label based indexing)
s.loc["Cherry"]   # Here we got the same 30 calories of the Cherry..
```

```python
s.loc[["Apple","Mango"]]    # Multiple Labeled based elements can also be access..
```

```python
s['Banana':'Mango']       # SLICING in Labeled indices are included both indices of start and end like here Mango is included..
```

## 2. Dictionary based Series

```python
fruit_protein = {
    "Avocado": 2.0,     # grams of protein
    "Guava": 2.6,
    "Blackberries": 2.0,
    "Oranges": 0.9,
    "Banana": 1.1,
    "Apples": 0.3,
    "Kiwi": 1.1,
    "Pomegranate": 1.7,
    "Mango": 0.8,
    "Cherries": 1.0
}
```

```python
s2 = pd.Series(fruit_protein, name="Protein")
s2
```

```python
s2.name = "Fruit Protein"
```

```python
s2
```

```python
s2.name
```

```python
s2.dtype
```

```python
s2.index
```

```python
s2.values
```

## 3. Conditional Selection

```python
# Suppose we need the fruits which has protein > 1
s2>1    # Whichever fruit has larger than 1 protein it will set as True and all other set as False
```

```python
# Now we want only fruits rows which satisfy this condition
s2[s2>1]
```

## 4. Logical Operators : and, or, not

```python
# Suppose we need fruits which have proteins >0.5 and <2
[(s2>.5) & (s2<2)]
```

```python
s2[(s2>.5) & (s2<2)]  # AND Operator use case 
```

```python
# For OR operator we will get the values which are >.5 or <2
s2[(s2>.5) | (s2<2)]  # OR Operator use case 
```

```python
len(s2)
```

```python
(s2>1)
```

```python
~(s2>1)
```

```python
s2[~(s2>1)]
```
