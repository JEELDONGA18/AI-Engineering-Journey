```python
import numpy as np
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve", "Alice"], 
    "Age": [25, 30, 35, np.nan, 29, 25],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "HR"], 
    "Salary": [50000, 60000, 70000, 62000, np.nan, 50000]
} 

df = pd.DataFrame(data)
df
```

```python
df["Promoted Salary"] = df["Salary"] * 10
```

```python
df
```

### 

# Data Cleaning

## 1.   Checking count of Null Values in df

```python
df.isnull().sum()
```

### 

## 2.  To remove rows which have null values

```python
df.dropna()
```

### Parameters of dropna()    --> How want our null values will be deletetd or drop

### A. how = "any"   (default)

```python
df.dropna(how = "any")  # any row that had any null value
```

### B. how = "all"

```python
df.dropna(how = "all") # If all the values in any row are null then we drop otherwise we do not touch it 
```

```python
df
```

### 

## 3. Fill the missing values

### A. Fill by some universal value like 0

```python
 df.fillna(0)  # This is noob method because we don't want the Salary and Age from nulll to 0 because that is not fair value
```

### B. Replace by mean

```python
df["Age"].fillna(df["Age"].mean())
```

```python
df
```

### C. Replace by Median

```python
df["Salary"].fillna(df["Salary"].median())
```

### D. Replace by Forward Fill --> Means the missing value's above value is present then it will be set as new value insted of missing value

```python
# df["Age"].fillna(method="ffill")  # --> Not working syntax
df["Age"].ffill()
```

### E. Replace by Backward Fill --> Means the missing value's below value is present then it will be set as new value insted of missing value 

```python
df["Age"].bfill()
```

## 5. Handling of duplicate value

### A. Identifying Duplicate Values

```python
df_dup = df[df.duplicated()]
df_dup
```

### 

### Woking of Duplicate checking

### 

#### Parameters of duplicated() 

#### I. keeps = "first" --> By default  --> This means mark other values except first as duplicated

```python
df_dup = df[df.duplicated(keep="first")]
df_dup
```

#### II. keeps = "last"  --> This means mark other values except last as duplicated

```python
df_dup = df[df.duplicated(keep="last")]
df_dup
```

### 

### B. Drop Duplicated value

```python
import numpy as np
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve", "Alice"], 
    "Age": [25, 30, 35, np.nan, 29, 25],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "HR"], 
    "Salary": [50000, 60000, 70000, 62000, np.nan, 50000]
} 

df = pd.DataFrame(data)
df
```

```python
df = df.drop_duplicates()
```
