# DataFrame

```python
import numpy as np
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve", "Alice"], 
    "Age": [25, 30, 35, np.nan, 29, 25],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "HR"], 
    "Salary": [50000, 60000, 70000, 62000, np.nan, 50000]
} 
```

```python
data
```

```python
import pandas as pd
df = pd.DataFrame(data)
df
```

### 

### To show dataframe 

```python
# To show first n rows 
df.head(2)   # By default value is 5 like df.head() --> will show you first 5 rows
```

```python
# To show last n rows
df.tail(3)   # By default value is 5 like df.tail() --> will show you last 5 rows
```

### 

###  iloc and loc

```python
# To show 1st and 2nd row only 
df.iloc[1:3]
```

```python
# Suppose want 1st and 2nd row with Age and Department column only then...
df.loc[1:3,["Age","Department"]]
```

```python
df.loc[1:3,["Age":"Salary"]]    # This experiment gives an error
```

```python
# To want this columns use column number with 'iloc' like
df.iloc[1:3, 1:4]    # range(rows), range(columns)  --> here end no of row is included but end no of column is excluded.
```
