```python
import numpy as np
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve", "Alice"], 
    "Age": [25, 30, 35, np.nan, 29, 25],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "HR"], 
    "Salary": [50000, 60000, 70000, 62000, np.nan, 50000]
} 

df = pd.DataFrame(data)
df
```

### 

```python
df["Promoted Salary"] = df["Salary"] * 10
```

```python
df
```

### 

# Dealing with Invalid Values

## A. Lambda Function

```python
# Just want to reduce the salary by 10th part whose salary is greater than 60,000 From Promoted Salary
df["Promoted Salary"] = df["Promoted Salary"].apply(lambda x: x/10 if x>60000 else x)
```

```python
df
```

### 

## B.  Split the Values

```python
df["Name"] = df["Name"].replace("Alice","Alice_Fernandes")
```

```python
df
```

```python
df[["First_Name","Last_Name"]] = df["Name"].str.split("_")
```

### 

## C. apply() usage 

```python
# Suppose we need to multiply all person's age by 2 but without using lambda function then..
def multiplying_by_two(x):
    return x*2

df["Age"] = df["Age"].apply(multiplying_by_two)   
```

```python
df
```

### 

```python
# Suppose we need to divide the age by 2 then..
df["Age"] = df["Age"].apply(lambda x:x/2)
```

```python
df
```

# Joints and Merges

```python
department_info = {
    "Department": ["HR", "IT", "Finance"],
    "Location": ["New York", "San Francisco", "Chicago"],
    "Manager": ["Laura", "Steve", "Nina"]
}

df2 = pd.DataFrame (department_info)
df2
```

### 

## A. concat()

```python
pd.concat([df,df2])  
```

```python
# To make this perfectly we can also do the vertical concatination also 
```

```python
pd.concat([df,df2], axis=1)
```

### 

## B. merge()

```python
# Here in above output there is Department column is repeats twice and we do not want it and so for it...
```

```python
pd.merge(df,df2, on="Department")
```
