# 3. Array Reshaping --> Reshape, Ravel, Flatten

## Reshape  --> Used to reshape the array dimension without affecting elements.

```python
arr8 = np.array([1,2,3,4,5,6])
arr8
```

```python
arr8_reshaped = arr8.reshape(2,3)
arr8_reshaped
```

```python
arr8_REreshaped = arr8_reshaped.reshape(3,2)
arr8_REreshaped
```

```python
arr8_1D = arr8_REreshaped.reshape(-1)
arr8_1D
```

### 

## Ravel --> Used to convert nD array to 1D array.

```python
arr9_ravel = arr8_REreshaped.ravel()
arr9_ravel
```

```python
arr9_ravel[0] = 100 
arr9_ravel
```

```python
arr8_REreshaped
```

### 

## Flatten --> Used to convert nD array to 1D array.

```python
arr10_flatten = arr8_REreshaped.flatten()
arr10_flatten
```

```python
arr10_flatten[0] = 1
```

```python
arr10_flatten
```

```python
arr8_REreshaped
```
